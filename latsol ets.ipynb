```python
##### a. input variabel bulan apa akan menginap
bulan = input("Bulan menginap: ")

# b. input variabel berapa hari akan menginap
lama = int(input("Lama menginap (hari): "))

# c. 
if bulan in ["Mei", "Oktober"]:
    studio = 500000
    apartment = 650000
elif bulan in ["Juni", "September"]:
    studio = 752000
    apartment = 687000
elif bulan in ["Juli", "Agustus"]:
    studio = 760000
    apartment = 770000
else:
    print("Bulan tidak valid.")

total_studio = studio * lama
total_apartment = apartment * lama

if lama > 3:
    total_studio == total_studio * 0.15
    total_apartment == total_apartment * 0.10

print(f"Studio: {total_studio:,.0f}")
print(f"Apartment: {total_apartment:,.0f}")
```

```python
# a. Menghapus angka atau huruf yang muncul berulang
Data = [2, 14, 2, 10, 5, 6, 2]
Hasil = []
for k in Data:
    if Data.count(k) == 1:
        Hasil.append(k)
print("a. Output =", Hasil)

# b. Kata terpanjang dalam kalimat
Kalimat = "menjadi mahasiswa yang dapat mengelola waktu dengan lebih baik"
Kata_terpanjang = ""
for kata in Kalimat.split():
    if len(kata) > len(Kata_terpanjang):
        Kata_terpanjang = kata
print("b. Output =", Kata_terpanjang)

# c. Menambah semua angka yang ada dalam list
Input_c = ["aku", 13, "ok", 55, "lama", 10]
Jumlah = 0
for d in Input_c:
    if isinstance(d, (int, float)):
        Jumlah += d
print("c. Output =", Jumlah)
```

```python
# Data pengiriman
data_pengiriman = (
    ("Laptop", 2.5, "Surabaya", 35000),
    ("Printer", 6.0, "Jakarta", 55000),
    ("Monitor", 4.2, "Bandung", 45000),
    ("Keyboard", 1.2, "Surabaya", 30000),
    ("Mouse", 0.5, "Jakarta", 25000)
)

# a. Seluruh data pengiriman ke kota Surabaya
print("a. Data pengiriman ke kota Surabaya:")
for data in data_pengiriman:
    if data[2] == "Surabaya":
        print(data)
print()

# b. Total berat seluruh barang yang dikirim ke Jakarta
total_berat = 0
for data in data_pengiriman:
    if data[2] == "Jakarta":
        total_berat += data[1]
print("b. Total berat barang ke Jakarta =", total_berat, "kg")
print()

# c. Pengiriman ke Bandung naik 10%
data_baru = []
for data in data_pengiriman:
    if data[2] == "Bandung":
        biaya_baru = data[3] * 1.10
        data_baru.append((data[0], data[1], data[2], biaya_baru))
    else:
        data_baru.append(data)

print("c. Tuple hasil setelah biaya Bandung naik 10%:")
for d in data_baru:
    print(d)
```

```python
# Menghitung akar polinomial
a, b, c = 4, 10, 4
D = (b**2 - 4*a*c)**0.5

# a
root1 = (-b + D) / (2*a)
# b
root2 = (-b - D) / (2*a)

print("a. Root 1:", root1)
print("b. Root 2:", root2)

# c
diff = []
for x in range(1,11):
    fx = 2*a*x+b
    diff.append(fx)

print("c. f\'(x) untuk x=1,...,10:", diff)
```

```python
# a. Input dari pengguna
barang = input("Nama barang: ")
jumlah_awal = float(input("Jumlah awal: "))
satuan_awal = input("Satuan awal (pcs/lusin/kodi): ").lower()
satuan_tujuan = input("Satuan tujuan (pcs/lusin/kodi): ").lower()

# satuan konversi
lusin_pcs = 12
kodi_pcs = 20

# b. Konversi satuan awal ke pcs
if satuan_awal == "pcs":
    total_pcs = jumlah_awal
elif satuan_awal == "lusin":
    total_pcs = jumlah_awal * lusin_pcs
elif satuan_awal == "kodi":
    total_pcs = jumlah_awal * kodi_pcs
else:
    print("Error: satuan awal tidak dikenal")

# c. Konversi pcs ke satuan tujuan
if satuan_tujuan == "pcs":
    hasil_konversi = total_pcs
elif satuan_tujuan == "lusin":
    hasil_konversi = total_pcs / lusin_pcs
elif satuan_tujuan == "kodi":
    hasil_konversi = total_pcs / kodi_pcs
else:
    print("Error: satuan tujuan tidak dikenal")

print("Barang           :", barang)
print(f"Jumlah awal     : {jumlah_awal} {satuan_awal}")
print(f"Total dalam pcs : {total_pcs:.0f} pcs")
print(f"Konversi ke {satuan_tujuan} : {hasil_konversi:.2f} {satuan_tujuan}")
```

```python
nilai=[78, 85, 90, 65, 85, 72, 90, 55, 78, 95, 60]
total=0
banyak=0
for n in nilai:
    total=total+n
    banyak=banyak+1
rata_rata=total/banyak
print("Rata-rata", rata_rata)

jumlah_di_atas=0
for n in nilai:
    if n> rata_rata:
        jumlah_di_atas=jumlah_di_atas+1
print("Nilai di atas rata-rata:", jumlah_di_atas)

nilai_unik=[]
for n in nilai:
    if n>=75 and n not in nilai_unik:
        nilai_unik.append(n)
print("Nilai unik >=75:", nilai_unik)
```
